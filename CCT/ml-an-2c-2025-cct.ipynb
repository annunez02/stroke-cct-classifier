```python
# ====================== IMPORTS ======================
import os
import random
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader, Subset
from torchvision import transforms
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score, precision_recall_curve, auc
from sklearn.model_selection import train_test_split
from PIL import Image
from tqdm import tqdm
import cv2
import json

# ====================== CONFIG ======================
BASE_PATH = '/kaggle/input/brain-stroke-ct-dataset/Brain_Stroke_CT_Dataset'
BATCH_SIZE = 16  
IMG_SIZE = 256   # High Def
NUM_EPOCHS = 24
LR = 1e-4
NUM_WORKERS = 4
SEED = 42
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
SAVE_DIR = '/kaggle/working'
os.makedirs(SAVE_DIR, exist_ok=True)

# Reproducibility
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)


# ====================== DATA AUGMENTATION ======================

train_transform = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomAffine(degrees=10, translate=(0.05, 0.05), scale=(0.95, 1.05)),
    transforms.ToTensor(),
    transforms.Normalize((0.5,), (0.5,))
])

val_transform = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize((0.5,), (0.5,))
])

# ====================== DATASET (RAW GRAYSCALE) ======================
class StandardCTDataset(Dataset):
    # store file paths and labels.
    def __init__(self, samples, labels, transform=None):
        self.samples = samples
        self.labels = labels
        self.transform = transform

    def __len__(self):
        return len(self.samples)

    #called when loader asks for a batch
    def __getitem__(self, idx):
        path = self.samples[idx]
        img = Image.open(path).convert("L") # Luminance (grayscale)
        
        if self.transform:
            img = self.transform(img)
        
        filename = path # for visualization
        return img, self.labels[idx], filename

# list of file paths and numeric labels 
def get_samples_labels(base_path, class_map): 
    samples, labels = [], []
    
    for cls, label_idx in class_map.items():
        folder = os.path.join(base_path, cls, 'PNG')
        if not os.path.exists(folder): continue
        
        for f in os.listdir(folder):
            if f.lower().endswith('.png'):
                samples.append(os.path.join(folder, f))
                labels.append(label_idx)
    
    return samples, labels

# split data into train (80%) and val(20%) - train_test_split
def make_loaders(base_path, classes_dict, val_split=0.2):
    samples, labels = get_samples_labels(base_path, classes_dict)
    
    train_idx, val_idx = train_test_split(
        np.arange(len(labels)), test_size=val_split, 
        stratify=labels, random_state=SEED, shuffle=True)
    
    # Create Subsets manually to pass to custom Dataset
    train_ds = StandardCTDataset([samples[i] for i in train_idx], [labels[i] for i in train_idx], transform=train_transform)
    val_ds = StandardCTDataset([samples[i] for i in val_idx], [labels[i] for i in val_idx], transform=val_transform)
    
    train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True, num_workers=NUM_WORKERS, pin_memory=True)
    val_loader = DataLoader(val_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS, pin_memory=True)
    
    return train_loader, val_loader

# ====================== HIGH-DEF MODEL ======================

# modifico el layer del transformer para que me devuelva weights que necesito para la visualizacion
class WeightsEnabledEncoderLayer(nn.TransformerEncoderLayer):
    def _sa_block(self, x, attn_mask, key_padding_mask, is_causal=False):
        x, weights = self.self_attn(
            x, x, x, attn_mask=attn_mask, 
            key_padding_mask=key_padding_mask, 
            need_weights=True, # ensures the attention map is calculated and returned
            is_causal=is_causal)
        
        return self.dropout1(x)


# Una imagen 256x256 flattened es 65,536 pixeles, demasiado grande para un transformer, con capas conv paso de 256 - 128 - 64 a 32 
# Entonces el transformer usa 32×32=1024 tokens, que es mas eficiente
class CompactConvTransformer(nn.Module):
    def __init__(self, num_classes=2, img_size=256, in_channels=1, emb_dim=64, n_heads=4, depth=2):
        super().__init__()
        
        # Conv Stem with MaxPool for 256 -> 32 reduction
        self.conv_stem = nn.Sequential(
            nn.Conv2d(in_channels, 32, 3, stride=2, padding=1),
            nn.ReLU(),
            nn.Conv2d(32, 64, 3, stride=2, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=2, stride=2), 
            nn.Dropout(0.1)
        )
        
        reduced = img_size // 8 
        self.seq_len = reduced * reduced + 1
        
        self.embedding = nn.Linear(64, emb_dim) # proyecta  CNN features a un vector size of 64
        self.cls_token = nn.Parameter(torch.randn(1, 1, emb_dim)) # model aggregates all information into this single token to make the final classification decision
        self.pos_embedding = nn.Parameter(torch.randn(1, self.seq_len, emb_dim)) # information about where each patch is in the image
        # los transformers no "saben" up/down/left/right
        
        # Transformer layers (Self-Attention) that learn global relationships between different parts of the brain scan
        encoder_layer = WeightsEnabledEncoderLayer(d_model=emb_dim, nhead=n_heads, dim_feedforward=128, dropout=0.1, batch_first=True)
        self.encoder = nn.TransformerEncoder(encoder_layer, num_layers=depth)
        
        self.norm = nn.LayerNorm(emb_dim)
        self.head = nn.Linear(emb_dim, num_classes) #final Linear layer

    def forward(self, x):
        x = self.conv_stem(x)
        x = x.flatten(2).transpose(1, 2)
        x = self.embedding(x)
        
        cls_tokens = self.cls_token.expand(x.shape[0], -1, -1)
        x = torch.cat((cls_tokens, x), dim=1)
        x = x + self.pos_embedding
        x = self.encoder(x)
        
        x = x[:, 0]
        x = self.norm(x)
        return self.head(x)

        
# ====================== TRAINING UTILS ======================

def train_epoch(model, loader, opt, crit, scaler):
    model.train()
    total_loss = 0
    correct = 0
    total_samples = 0
    
    for img, lab, _ in tqdm(loader, desc="Train", leave=False):
        img, lab = img.to(DEVICE), lab.to(DEVICE)
        opt.zero_grad(set_to_none=True)
        
        # Mixed Precision (float16) to speed up training and reduce memory usage on the GPU
        with torch.amp.autocast('cuda', enabled=(DEVICE.type=='cuda')):
            logits = model(img)
            loss = crit(logits, lab)
        
        scaler.scale(loss).backward()
        scaler.step(opt)
        scaler.update()
        
        # Calculate Metrics
        total_loss += loss.item() * img.size(0)
        preds = logits.argmax(dim=1)
        correct += (preds == lab).sum().item()
        total_samples += lab.size(0)
    
    avg_loss = total_loss / len(loader.dataset)
    acc = correct / total_samples
    
    return avg_loss, acc

def validate(model, loader, crit):
    model.eval()
    total_loss = 0
    preds, trues, probs = [], [], []
    
    with torch.no_grad():
    
        for img, lab, _ in loader:
            img, lab = img.to(DEVICE), lab.to(DEVICE)
            logits = model(img)
            loss = crit(logits, lab)
            total_loss += loss.item() * img.size(0)
            preds.extend(logits.argmax(1).cpu().numpy())
            trues.extend(lab.cpu().numpy())
            probs.extend(torch.softmax(logits, 1).cpu().numpy())
    
    return total_loss / len(loader.dataset), np.array(trues), np.array(preds), np.array(probs)

def run_training(model_name, classes_map):
    print(f"\n--- Training {model_name} (High Def 256) ---")
    train_loader, val_loader = make_loaders(BASE_PATH, classes_map)
    
    # Class Weights, calcula pesos para balancear loss
    train_labels = train_loader.dataset.labels
    counts = np.bincount(train_labels)
    weights = torch.tensor([sum(counts)/(2*c) for c in counts], dtype=torch.float).to(DEVICE)
    print(f"Class Weights: {weights.cpu().numpy()}")

    model_config = {
        "num_classes": 2,
        "img_size": IMG_SIZE,
        "in_channels": 1, 
        "emb_dim": 64, 
        "n_heads": 4, 
        "depth": 2
    }
    
    with open(os.path.join(SAVE_DIR, f"{model_name}_config.json"), 'w') as f:
        json.dump(model_config, f)
    with open(os.path.join(SAVE_DIR, f"{model_name}_classes.json"), 'w') as f:
        json.dump(classes_map, f)
    
    model = CompactConvTransformer(**model_config).to(DEVICE)
    opt = torch.optim.AdamW(model.parameters(), lr=LR, weight_decay=1e-3) # Adam with Weight Decay, standard transformers
    crit = nn.CrossEntropyLoss(weight=weights, label_smoothing=0.1) # avoid overfitting
    scaler = torch.amp.GradScaler('cuda', enabled=(DEVICE.type=='cuda'))
    
    best_loss = float('inf')
    
    # === METRICS HISTORY STORAGE ===
    history = {'train_loss': [], 'train_acc': [], 'val_loss': [], 'val_acc': []}

    for epoch in range(NUM_EPOCHS):
        tl, ta = train_epoch(model, train_loader, opt, crit, scaler) # NOW returns accuracy too
        vl, yt, yp, _ = validate(model, val_loader, crit)
        va = accuracy_score(yt, yp)
        
        print(f"Ep {epoch+1}: Train Loss {tl:.4f} | Train Acc {ta:.4f} | Val Loss {vl:.4f} | Val Acc {va:.4f}")
    
        # Store metrics
        history['train_loss'].append(tl)
        history['train_acc'].append(ta)
        history['val_loss'].append(vl)
        history['val_acc'].append(va)

        if vl < best_loss:
            best_loss = vl
            torch.save(model.state_dict(), os.path.join(SAVE_DIR, f"{model_name}.pt"))
    
    plt.figure(figsize=(12, 5))
    
    # Plot Loss
    plt.subplot(1, 2, 1)
    plt.plot(history['train_loss'], label='Train Loss')
    plt.plot(history['val_loss'], label='Val Loss')
    plt.title(f'{model_name} Loss History')
    plt.xlabel('Epochs')
    plt.ylabel('Loss')
    plt.legend()
    
    # Plot Accuracy
    plt.subplot(1, 2, 2)
    plt.plot(history['train_acc'], label='Train Accuracy', color='blue')
    plt.plot(history['val_acc'], label='Val Accuracy', color='green')
    plt.title(f'{model_name} Accuracy History')
    plt.xlabel('Epochs')
    plt.ylabel('Accuracy')
    plt.legend()
    
    plt.show()

    return model, val_loader

# ====================== EXPLANABILITY ======================
# Gradient-weighted Class Activation Mapping, para ver que regiones eran las mas activas en la ultima conv layer en la prediccion
class GradCAM: 
    def __init__(self, model, target_layer):
        self.model = model
        self.gradients = None
        self.activations = None
        target_layer.register_forward_hook(self.save_activation)
        target_layer.register_full_backward_hook(self.save_gradient)

    def save_activation(self, module, input, output): self.activations = output.detach()
    def save_gradient(self, module, grad_input, grad_output): self.gradients = grad_output[0].detach()

    def __call__(self, x):
        self.model.zero_grad()
        logits = self.model(x)
        class_idx = logits.argmax(dim=1).item()
        loss = logits[0, class_idx]
        loss.backward(retain_graph=True)
        
        weights = self.gradients.mean(dim=(2,3), keepdim=True)
        cam = (weights * self.activations).sum(dim=1, keepdim=True)
        cam = torch.relu(cam).squeeze().cpu().numpy()
        cam = (cam - cam.min()) / (cam.max() - cam.min() + 1e-8)
        return cam, logits, torch.softmax(logits, dim=1)

# Hooks into the Transformer layers to extract the Self-Attention maps
class AttentionExtractor:
    def __init__(self, model):
        self.model = model
        self.attns = []
        for layer in self.model.encoder.layers:
            layer.self_attn.register_forward_hook(self.capture)
    
    def capture(self, module, inp, out):
        if isinstance(out, tuple) and out[1] is not None: self.attns.append(out[1].detach().cpu())
    
    def __call__(self, x):
        self.attns = []
        _ = self.model(x)
        return self.attns

# Combines attention maps across layers to show how information flows from the pixels to the final classification token
def compute_rollout(attns):
    if not attns: return None
    result = None
    
    for attn in attns:
        a = attn[0].mean(dim=0) # Batch 0, Mean Heads
        a = a + torch.eye(a.shape[0])
        a = a / a.sum(dim=-1, keepdim=True)
        if result is None: result = a
        else: result = torch.matmul(a, result)
    return result[0, 1:] # CLS row

    
# ====================== EVALUATION PLOTS ======================

# load best saved model
def evaluate_performance(model, loader, model_name, class_names, threshold=None):
    model.load_state_dict(torch.load(os.path.join(SAVE_DIR, f"{model_name}.pt")))
    model.eval()
    
    _, y_true, y_pred, y_prob = validate(model, loader, nn.CrossEntropyLoss())
    
    if threshold:
        print(f"\n[INFO] Applying {threshold*100}% Confidence Threshold for Class 'Normal'")
        y_pred = []
        
        for p in y_prob:
            y_pred.append(0 if p[0] > threshold else 1)
        
        y_pred = np.array(y_pred)

    print(f"\n=== {model_name} REPORT ===")
    print(classification_report(y_true, y_pred, target_names=class_names))
    
    # Confusion Matrix
    cm = confusion_matrix(y_true, y_pred)
    plt.figure(figsize=(5,4))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=class_names, yticklabels=class_names)
    plt.title(f"{model_name} Confusion Matrix")
    plt.show()
    
    # PR Curve
    precision, recall, _ = precision_recall_curve(y_true, y_prob[:, 1])
    pr_auc = auc(recall, precision)
    plt.figure(figsize=(6,4))
    plt.plot(recall, precision, label=f'AUC = {pr_auc:.2f}')
    plt.xlabel('Recall')
    plt.ylabel('Precision')
    plt.title(f"{model_name} PR Curve (Class: {class_names[1]})")
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.show()

# ====================== VISUALIZATION FUNCTION ======================

def visualize_hd(model_A, model_B, loader):
    model_A.eval()
    model_B.eval()
    
    cam_A = GradCAM(model_A, model_A.conv_stem[-2]) 
    att_ext_A = AttentionExtractor(model_A)
    
    cam_B = GradCAM(model_B, model_B.conv_stem[-2])
    att_ext_B = AttentionExtractor(model_B)
    
    val_iter = iter(loader)
    print("\nVisualizing High-Def Predictions (With Agreement Map)...")
    
    for _ in range(20):
        try: img, label, fname = next(val_iter)
        except: break
        
        img_tensor = img[0].unsqueeze(0).to(DEVICE)
        gt = label[0].item()
        
        # --- Stage A ---
        cam_map_A, logits_A, probs_A = cam_A(img_tensor)
        prob_normal = probs_A[0, 0].item()
        
        if prob_normal > 0.70: 
            stage = "A"
            pred_desc = "Normal"
            cam_map = cam_map_A
            attns = att_ext_A(img_tensor)
        else:
            # --- Stage B ---
            stage = "B"
            cam_map_B, logits_B, probs_B = cam_B(img_tensor)
            pred_b = logits_B.argmax(1).item()
            cond = "Bleeding" if pred_b == 0 else "Ischemia"
            pred_desc = f"Abnormal ({cond})"
            cam_map = cam_map_B
            attns = att_ext_B(img_tensor)

        # Attention Map
        rollout = compute_rollout(attns)
        if rollout is not None:
            # Reshape 1024 -> 32x32 (since 256/8 = 32)
            attn_map = rollout.view(32, 32).numpy()
            attn_map = cv2.resize(attn_map, (IMG_SIZE, IMG_SIZE))
            attn_map = (attn_map - attn_map.min()) / (attn_map.max() - attn_map.min() + 1e-8)
        else:
            attn_map = np.zeros((IMG_SIZE, IMG_SIZE))
            
        cam_map = cv2.resize(cam_map, (IMG_SIZE, IMG_SIZE))
        
        # --- AGREEMENT MAP ---
        combined_map = cam_map * attn_map
        combined_map = (combined_map - combined_map.min()) / (combined_map.max() - combined_map.min() + 1e-8)
        
        
        disp_img = img_tensor.cpu().squeeze().numpy()
        # Percentile norm to fix contrast
        p1, p99 = np.percentile(disp_img, 1), np.percentile(disp_img, 99)
        disp_img = np.clip((disp_img - p1) / (p99 - p1 + 1e-8), 0, 1)

        path_str = fname[0]
        if "Normal" in path_str:
            true_desc = "Normal"
        elif "Bleeding" in path_str:
             true_desc = "Abnormal (Bleeding)"
        elif "Ischemia" in path_str:
             true_desc = "Abnormal (Ischemia)"
        else:
             true_desc = "Abnormal (Unknown)"

        print(f"File: {os.path.basename(path_str)}") 
        print(f"True: {true_desc} | Pred: {pred_desc} | Prob Normal: {prob_normal:.2f}")
        
        # --- Plots ---
        plt.figure(figsize=(18, 4))
        
        # Input
        plt.subplot(1, 4, 1); plt.title("Input"); plt.imshow(disp_img, cmap='gray'); plt.axis('off')

        # GradCAM
        plt.subplot(1, 4, 2); plt.title("CNN (Grad-CAM)"); plt.imshow(disp_img, cmap='gray'); 
        plt.imshow(cam_map, cmap='jet', alpha=0.5)
        plt.axis('off')
        
        # Attention
        plt.subplot(1, 4, 3); plt.title("Transformer (Attn)"); plt.imshow(disp_img, cmap='gray'); 
        plt.imshow(attn_map, cmap='jet', alpha=0.5)
        plt.axis('off')
        
        # Agreement
        plt.subplot(1, 4, 4); plt.title("Combined Agreement"); plt.imshow(disp_img, cmap='gray'); 
        plt.imshow(combined, cmap='inferno', alpha=0.6)
        plt.axis('off')
        
        plt.tight_layout()
        plt.show()

# ====================== EXECUTION ======================

# Train
model_A, loader_A = run_training('Model_A_HD', {'Normal': 0, 'Bleeding': 1, 'Ischemia': 1})
evaluate_performance(model_A, loader_A, 'Model_A_HD', ['Normal', 'Abnormal'], threshold=0.70)

model_B, loader_B = run_training('Model_B_HD', {'Bleeding': 0, 'Ischemia': 1})
evaluate_performance(model_B, loader_B, 'Model_B_HD', ['Bleeding', 'Ischemia'])

# Visualize
visualize_hd(model_A, model_B, loader_A)
```

```python
# ====================== EXTERNAL TEST DATASET ======================

class ExternalTestDataset(Dataset):
    def __init__(self, root_dir, transform=None):
        self.root_dir = root_dir
        self.png_dir = os.path.join(root_dir, 'PNG')
        self.transform = transform
        
        csv_path = os.path.join(root_dir, 'labels.csv')
        
        try:
            self.df = pd.read_csv(csv_path)
            if 'image_id' not in self.df.columns: # por si era con ;
                self.df = pd.read_csv(csv_path, sep=';')
        except:
             print("Error reading CSV. Check format.")

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        
        fname = str(row['image_id'])
        label = int(row['Stroke']) # 0 or 1
        
        if not fname.lower().endswith('.png'):
            fname += '.png'
            
        img_path = os.path.join(self.png_dir, fname)
        
        # Safety check if file exists
        if not os.path.exists(img_path):
            # Return a black image and dummy label if file missing (prevents crash)
            return torch.zeros((1, IMG_SIZE, IMG_SIZE)), label, fname
            
        img = Image.open(img_path).convert("L")
        
        if self.transform:
            img = self.transform(img)
            
        return img, label, fname

# ====================== EXTERNAL EVALUATION LOGIC ======================

def evaluate_external_set(model_A, model_B):
    print("\n" + "="*40)
    print("   RUNNING EXTERNAL TEST EVALUATION")
    print("="*40)
    
    ext_path = '/kaggle/input/brain-stroke-ct-dataset/Brain_Stroke_CT_Dataset/External_Test'
    
    # Create Loader
    ds = ExternalTestDataset(ext_path, transform=val_transform)
    loader = DataLoader(ds, batch_size=BATCH_SIZE, shuffle=False)
    
    model_A.eval()
    model_B.eval()
    
    # Lists for Metrics
    y_true_A = [] # 0=No Stroke, 1=Stroke (From CSV)
    y_pred_A = [] # 0=Normal, 1=Abnormal (From Model A)
    
    # Counters for Model B (Since we don't know truth, we just count predictions)
    b_predictions = {'Bleeding': 0, 'Ischemia': 0}
    
    print(f"Processing {len(ds)} external images...")
    
    with torch.no_grad():
        for img, label, _ in tqdm(loader, desc="External Test"):
            img = img.to(DEVICE)
            label = label.cpu().numpy()
            
            # --- MODEL A (Detection) ---
            logits_A = model_A(img)
            probs_A = torch.softmax(logits_A, dim=1)
            
            # Batch processing
            for i in range(len(label)):
                p_norm = probs_A[i, 0].item()
                
                # Prediction Logic: If prob_normal > 0.70 -> 0 (Normal), else 1 (Stroke)
                pred_class_A = 0 if p_norm > 0.70 else 1
                
                y_true_A.append(label[i])
                y_pred_A.append(pred_class_A)
                
                # --- MODEL B (Classification) ---
                # Only ask Model B if Model A detects a stroke
                if pred_class_A == 1:
                    img_single = img[i].unsqueeze(0) 
                    logits_B = model_B(img_single)
                    pred_B = logits_B.argmax(1).item()
                    
                    # Mapping: 0=Bleeding, 1=Ischemia
                    if pred_B == 0: b_predictions['Bleeding'] += 1
                    else: b_predictions['Ischemia'] += 1

    # --- REPORTING ---
    
    print("\n" + "-"*30)
    print("MODEL A REPORT (Detection vs Ground Truth)")
    print("-"*30)
    # 0: Normal, 1: Stroke
    print(classification_report(y_true_A, y_pred_A, target_names=['Normal', 'Stroke']))
    
    cm = confusion_matrix(y_true_A, y_pred_A)
    plt.figure(figsize=(6, 5))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=['Normal', 'Stroke'], yticklabels=['Normal', 'Stroke'])
    plt.xlabel('Predicted')
    plt.ylabel('Actual')
    plt.title('Confusion Matrix: External Test')
    plt.show()

    print("\n" + "-"*30)
    print("MODEL B REPORT (Predictions Only)")
    print("-"*30)
    print("Note: Ground truth for Stroke Type (Bleeding/Ischemia) is not available.")
    print("Showing what Model B predicted for the detected strokes of Model A:")
    print(f"  - Predicted Bleeding: {b_predictions['Bleeding']}")
    print(f"  - Predicted Ischemia: {b_predictions['Ischemia']}")
    
    return ds

# ====================== EXTERNAL VISUALIZATION ======================

def visualize_external(model_A, model_B, dataset):
    indices = random.sample(range(len(dataset)), 10)
    
    model_A.eval()
    model_B.eval()

    cam_A = GradCAM(model_A, model_A.conv_stem[-2]) 
    att_ext_A = AttentionExtractor(model_A)
    
    cam_B = GradCAM(model_B, model_B.conv_stem[-2])
    att_ext_B = AttentionExtractor(model_B)
    
    print("\n" + "="*20 + " Visualizing External Samples " + "="*20)
    
    for idx in indices:
        img, true_label, fname = dataset[idx]
        img_tensor = img.unsqueeze(0).to(DEVICE)
        
        # --- Model A Prediction ---
        cam_map_A, logits_A, probs_A = cam_A(img_tensor)
        prob_norm = probs_A[0, 0].item()
        pred_A_idx = 0 if prob_norm > 0.70 else 1
        
        final_pred_str = "Normal"
        
        # Decide which map to show
        if pred_A_idx == 0:
            active_cam = cam_map_A
            active_attns = att_ext_A(img_tensor)
            model_used = "Model A"
        else:
            # --- Model B Prediction ---
            cam_map_B, logits_B, probs_B = cam_B(img_tensor)
            pred_B_idx = logits_B.argmax(1).item()
            
            subtype = "(Bleeding)" if pred_B_idx == 0 else "(Ischemia)"
            final_pred_str = f"Abnormal {subtype}"
            
            active_cam = cam_map_B
            active_attns = att_ext_B(img_tensor)
            model_used = "Model B"

        # Attention Map
        rollout = compute_rollout(active_attns)
        if rollout is not None:
            attn_map = rollout.view(32, 32).numpy()
            attn_map = cv2.resize(attn_map, (IMG_SIZE, IMG_SIZE))
            attn_map = (attn_map - attn_map.min()) / (attn_map.max() - attn_map.min() + 1e-8)
        else:
            attn_map = np.zeros((IMG_SIZE, IMG_SIZE))
            
        # Grad-CAM Map
        cam_map = cv2.resize(active_cam, (IMG_SIZE, IMG_SIZE))
        
        # COMBINED (Agreement Map)
        # Element-wise multiplication: Highlights only where BOTH agree
        combined_map = cam_map * attn_map
        combined_map = (combined_map - combined_map.min()) / (combined_map.max() - combined_map.min() + 1e-8)

        # --- IMAGE PROCESSING ---
        disp_img = img.squeeze().numpy()
        # Contrast fix for display
        p1, p99 = np.percentile(disp_img, 1), np.percentile(disp_img, 99)
        disp_img = np.clip((disp_img - p1) / (p99 - p1 + 1e-8), 0, 1)
        
        # --- TRUTH LABEL ---
        true_str = "Normal" if true_label == 0 else "Stroke (Unknown Type)"

        print(f"\nFile: {fname}")
        print(f"True: {true_str} | Pred: {final_pred_str} | (Prob Normal: {prob_norm:.2f})")
        print(f"Showing Explanation from: {model_used}")
        
        plt.figure(figsize=(16, 4))
        
        # Input
        plt.subplot(1, 4, 1); plt.title("Input"); plt.imshow(disp_img, cmap='gray'); plt.axis('off')
        
        # Grad-CAM (CNN)
        plt.subplot(1, 4, 2); plt.title("CNN Focus (Grad-CAM)"); plt.imshow(disp_img, cmap='gray')
        plt.imshow(cam_map, cmap='jet', alpha=0.5); plt.axis('off')
        
        # Attention (Transformer)
        plt.subplot(1, 4, 3); plt.title("Transformer Focus (Attn)"); plt.imshow(disp_img, cmap='gray')
        plt.imshow(attn_map, cmap='jet', alpha=0.5); plt.axis('off')
        
        # Plot 4: Combined (Agreement)
        plt.subplot(1, 4, 4); plt.title("Combined Agreement"); plt.imshow(disp_img, cmap='gray')
        plt.imshow(combined_map, cmap='inferno', alpha=0.6); plt.axis('off')
        
        plt.tight_layout()
        plt.show()

# ====================== EXECUTION ======================

# Run the external evaluation
ext_dataset = evaluate_external_set(model_A, model_B)

# Visualize random samples from it
visualize_external(model_A, model_B, ext_dataset)
```
